```python
import json
from pathlib import Path

jsonl_path = Path('/Users/sadi_/Coding/Small-Language-Model/All-Results/QWEN/SVAMP/results-notebook-13-piqa-three-angles-qwen-ft-svamp.jsonl')

records = []
if jsonl_path.exists() and jsonl_path.stat().st_size > 0:
    with open(jsonl_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                records.append(json.loads(line))
            except json.JSONDecodeError:
                continue  # skip malformed lines

df04 = pd.DataFrame(records)
run_name = 'PIQA-QWEN-FT-SVAMP.csv'
df04['fineTunedOn']='SVAMP'
df04['guideModel']='QWEN'
df04.to_csv('PIQA-QWEN-FT-SVAMP.csv')
```
